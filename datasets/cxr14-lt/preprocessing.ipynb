```python
import pandas as pd


train_df = pd.read_csv("nih-cxr-lt_single-label_train.csv")
test_df = pd.read_csv("nih-cxr-lt_single-label_test.csv")
val_df = pd.read_csv("nih-cxr-lt_single-label_balanced-val.csv")

train_df["Split"] = "TRAIN"
val_df["Split"] = "VAL"
test_df["Split"] = "TEST"

df = pd.concat([train_df, val_df, test_df])
df
```

```python
df = df.rename(columns={"id": "path", "subject_id": "id"})
df["path"] = df["path"].apply(lambda x: "images/" + x)
df
```

```python
df.to_csv("cxr14-lt.csv")
```
