```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datasets import datasets
```

```python
plt.figure(figsize=(20,15))

# Enable LaTeX font rendering
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern Roman"],
})


for i, (ds_name, ds) in enumerate(datasets.items(), 1):
    df = pd.read_csv(ds.filelist)
    df = df.drop(columns="Unnamed: 0", errors="ignore")
    meta_columns = ["id", "path", "Split"]
    
    # Get class columns by excluding meta columns
    class_columns = [col for col in df.columns if col not in meta_columns]
    
    if ds_name == "imagenet":
        # Randomly sample 40 classes for ImageNet
        class_columns = np.random.choice(class_columns, size=40, replace=False)
    
    # Calculate class distribution
    class_dist = df[class_columns].sum()
    
    # Sort values in descending order
    class_dist = class_dist.sort_values(ascending=False)
    
    # Create subplot
    plt.subplot(3, 2, i)
    sns.barplot(x=class_dist.index, y=class_dist.values)
    plt.title(f"Class Distribution - {ds.display_name} ({len(class_columns)} classes)")
    plt.xticks(rotation=45, ha='right')
    plt.ylabel("Count")

plt.tight_layout()
plt.savefig('class_distributions.pdf')
plt.show()
```

```python
#isic.id.unique()
import matplotlib.pyplot as plt
from PIL import Image
import os
import numpy as np

for ds_name, ds in datasets.items(): 
    if not ds.has_id: 
        continue
    np.random.seed(11)
    # Get 3 random IDs from training set
    data = pd.read_csv(ds.filelist)
    train_ids = data[data['Split'] == 'TRAIN']['id'].unique()

    # Set up the plot
    fig, axes = plt.subplots(3, 4, figsize=(15, 12))
    plt.suptitle(f"{ds.display_name}", fontsize=22)

    # Base directory for images
    data = data[data.id.map(data.id.value_counts())  > 3]

    selected_ids = np.random.choice(data.id.unique().tolist(), 3, replace=False)
    # For each selected ID
    for row, id_num in enumerate(selected_ids):
        # Get all samples for this ID, up to 4
        id_samples = data[data['id'] == id_num]['path']
        n_samples = min(4, len(id_samples))
        samples = id_samples.sample(n=n_samples).values
        
        # Plot each sample
        for col, img_path in enumerate(samples):
            img = Image.open(os.path.join(ds.basedir, img_path))
            img = img.resize((512, 512), Image.Resampling.LANCZOS)
            if ds.viz_grey_scale: 
                axes[row, col].imshow(img, cmap="Greys_r")
            else: 
                axes[row, col].imshow(img)
            axes[row, col].axis('off')
    #        if col == 0:  # Only add ID label to first image in row
    #            axes[row, col].set_title(f'ID: {id_num}')

    plt.tight_layout()
    plt.savefig(f"output/{ds_name}_samples.png")
    plt.show()
```
