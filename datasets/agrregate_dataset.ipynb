# Aggregate datasets with automatic oversampling 

```python
from datasets import datasets
import pandas as pd
dfs = [] 
total_ids = 0 
for ds_name, ds in datasets.items(): 
    df = pd.read_csv(ds.filelist)
    df = df[["path", "id", "Split"]]
    df["id"] = df["id"].apply(lambda x: x + total_ids)
    total_ids += df["id"].max().item()
    total_ids += 1 # just to be sure in case id starts with 0 
    df["ds_name"] = ds_name

    # Only oversample test split
    train_val_df = df[df["Split"] != "TEST"]
    test_df = df[df["Split"] == "TEST"]
    dfs.extend([train_val_df]*ds.oversample_factor)
    dfs.append(test_df)

df = pd.concat(dfs)
df.to_csv("alldatasets_trainvalbalanced.csv")
df.sample(20)
```
